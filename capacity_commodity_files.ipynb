```python
"""
split_by_commodity.py

Reads a CSV file and splits it into separate CSV files, one per commodity type.
Each output file is named after its commodity (e.g., Electricity.csv) and is
saved alongside the input file by default.

Usage:
    python split_by_commodity.py [input_file] [output_dir]

Arguments:
    input_file  : Path to the source CSV file (optional, defaults to the
                  results_004 capacity.csv path below)
    output_dir  : Directory to write output CSVs (optional, defaults to the
                  same directory as the input file)

Examples:
    python split_by_commodity.py
    python split_by_commodity.py /path/to/capacity.csv
    python split_by_commodity.py /path/to/capacity.csv ./output_csvs
"""

import sys
import os
import pandas as pd

DEFAULT_INPUT = (
    "/Users/abbie/MacroEnergy-Abbie.jl/MacroEnergyExamples/"
    "9Zone_US/all_168hr/"
    "results_001/results/capacity.csv"
)

def split_by_commodity(input_path: str, output_dir=None) -> None:
    # ── Validate input ──────────────────────────────────────────────────────
    if not os.path.isfile(input_path):
        print(f"Error: file not found — {input_path}")
        sys.exit(1)

    # ── Load the CSV ─────────────────────────────────────────────────────────
    print(f"Reading: {input_path}")
    df = pd.read_csv(input_path)

    if "commodity" not in df.columns:
        print("Error: CSV does not contain a 'commodity' column.")
        sys.exit(1)

    # ── Resolve output directory ──────────────────────────────────────────────
    if output_dir is None:
        output_dir = os.path.dirname(os.path.abspath(input_path))

    os.makedirs(output_dir, exist_ok=True)

    # ── Split and write ───────────────────────────────────────────────────────
    commodities = df["commodity"].unique()
    print(f"Found {len(commodities)} commodity type(s): {', '.join(commodities)}\n")

    for commodity in commodities:
        subset = df[df["commodity"] == commodity]

        # Make the filename filesystem-safe (replace spaces/slashes with underscores)
        safe_name = commodity.replace(" ", "_").replace("/", "_")
        out_path = os.path.join(output_dir, f"{safe_name}.csv")

        subset.to_csv(out_path, index=False)
        print(f"  Written {len(subset):>5} rows  →  {out_path}")

    print(f"\nDone. {len(commodities)} file(s) saved to: {output_dir}")


if __name__ == "__main__":
    split_by_commodity(DEFAULT_INPUT)
```
